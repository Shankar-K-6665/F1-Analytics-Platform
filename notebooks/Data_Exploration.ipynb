```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display all columns
pd.set_option('display.max_columns', None)

print("Libraries imported successfully!")
```

```python
# Load datasets

drivers = pd.read_csv("../data/raw/drivers.csv")
constructors = pd.read_csv("../data/raw/constructors.csv")
circuits = pd.read_csv("../data/raw/circuits.csv")
seasons = pd.read_csv("../data/raw/seasons.csv")

print("All datasets loaded successfully!")
```

```python
nationality = drivers["nationality"].value_counts().head(10)

plt.figure(figsize=(10,5))
nationality.plot(kind="bar")
plt.title("Top 10 Driver Nationalities")
plt.xlabel("Nationality")
plt.ylabel("Number of Drivers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
team_country = constructors["nationality"].value_counts()

plt.figure(figsize=(10,5))
team_country.plot(kind="bar")
plt.title("Constructors by Nationality")
plt.xlabel("Nationality")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
country = circuits["Location.country"].value_counts().head(15)

plt.figure(figsize=(12,6))
country.plot(kind="bar")
plt.title("Top Countries Hosting Formula 1 Circuits")
plt.xlabel("Country")
plt.ylabel("Number of Circuits")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
country = circuits["Location.country"].value_counts().head(15)

plt.figure(figsize=(12,6))
country.plot(kind="bar")
plt.title("Top Countries Hosting Formula 1 Circuits")
plt.xlabel("Country")
plt.ylabel("Number of Circuits")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14,4))
plt.plot(seasons["season"].astype(int), marker="o")
plt.title("Formula 1 Seasons")
plt.xlabel("Year")
plt.ylabel("Season")
plt.grid(True)
plt.show()
```
